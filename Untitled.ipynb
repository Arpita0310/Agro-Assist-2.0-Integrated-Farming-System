```python
import pandas as pd
import requests
import time

# Load your dataset
df = pd.read_excel("district_info.xlsx")

latitudes = []
longitudes = []

for index, row in df.iterrows():
    district = row["District"]
    state = row["State"]

    query = f"{district}, {state}, India"

    try:
        url = f"https://nominatim.openstreetmap.org/search?q={query}&format=json&limit=1"
        headers = {"User-Agent": "AgroAssistApp"}

        res = requests.get(url, headers=headers).json()

        if res:
            lat = res[0]["lat"]
            lon = res[0]["lon"]
        else:
            lat, lon = None, None

    except:
        lat, lon = None, None

    latitudes.append(lat)
    longitudes.append(lon)

    print(f"{district} → {lat}, {lon}")

    time.sleep(1)  # IMPORTANT (avoid API block)

# Add columns
df["Latitude"] = latitudes
df["Longitude"] = longitudes

# Save updated file
df.to_excel("district_info_updated.xlsx", index=False)

print("✅ Done! File saved.")
```
